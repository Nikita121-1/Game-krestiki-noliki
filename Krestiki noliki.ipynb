```python
def print_field(field):
    print(f"\n {' '.join(map(str, range(3)))}") # Печатает индексы столбцов: 0 1 2
    for i, row in enumerate(field):
        print(f"{i} {' '.join(row)}")
        print()

def ask_coords(field):
    while True:
        try:
            sel = input("Ваш ход (строка и столбец через пробел): ").split()

            if len(sel) != 2:
                print("Ошибка! Введите ДВА числа через пробел.")
                continue

            x, y = map(int, sel)

            if not (0 <= x <= 2 and 0 <= y <= 2):
                print("Ошибка! Координаты должны быть от 0 до 2.")
                continue

            if field[x][y] != "-":
                print("Ошибка! Эта клетка уже занята.")
                continue

            return x, y

        except ValueError:
            print("Ошибка! Введите корректные числа.")

def check_win(field, user):
# Горизонтали и вертикали
    for i in range(3):
        if all(field[i][j] == user for j in range(3)): return True
        if all(field[j][i] == user for j in range(3)): return True

# Диагонали
        if all(field[i][i] == user for i in range(3)): return True
        if all(field[i][2-i] == user for i in range(3)): return True

        return False

def start_game():
# Инициализация поля 3х3
    field = [["-"] * 3 for _ in range(3)]
    count = 0

    print("--- Игра КРЕСТИКИ-НОЛИКИ ---")

    while True:
        print_field(field)

# Определяю, чей сейчас ход
        current_user = "X" if count % 2 == 0 else "0"
        print(f"Ходит игрок: {current_user}")

        x, y = ask_coords(field)
        field[x][y] = current_user

        if check_win(field, current_user):
            print_field(field)
            print(f"ПОЗДРАВЛЯЕМ! Победил игрок: {current_user}!")
            break

            count += 1
            if count == 9:
                print_field(field)
                print("Ничья! Победила дружба.")
                break


if __name__ == "__main__":
    start_game()
```
