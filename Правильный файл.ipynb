```python
def show_field(f):
    print(f" 0 1 2")
    for i in range(3):
        row_info = " ".join(f[i])
        print(f"{i} {row_info}")

def check_win(f, user):

    for i in range(3):
        if all([f[i][0] == user, f[i][1] == user, f[i][2] == user]):
            return True
        if all([f[0][i] == user, f[1][i] == user, f[2][i] == user]):
            return True
# Проверка диагоналей
        if all([f[0][0] == user, f[1][1] == user, f[2][2] == user]):
            return True
        if all([f[0][2] == user, f[1][1] == user, f[2][0] == user]):
            return True
        return False

def start_game():
# Создаем пустое поле
    field = [["-"] * 3 for _ in range(3)]

# Очередность: True - ходит X, False - ходит 0
    x_turn = True
    count = 0

    print("--- Игра Крестики-нолики ---")

    while True:
        show_field(field)

# Определяем текущий символ
        current_user = "X" if x_turn else "0"
        print(f"Ход игрока: {current_user}")

# Ввод координат с проверкой
        try:
            line = input("Введите строку и столбец через пробел: ").split()
            if len(line) != 2:
                print("Ошибка! Введите 2 числа.")
                continue

            x, y = map(int, line)

            if not (0 <= x <= 2 and 0 <= y <= 2):
                print("Ошибка! Числа должны быть от 0 до 2.")
                continue

            if field[x][y] != "-":
                print("Ошибка! Клетка занята.")
                continue
        except ValueError:
            print("Ошибка! Введите целые числа.")
            continue

# Ставим символ на поле
        field[x][y] = current_user
        count += 1

# Проверяем на победу
        if check_win(field, current_user):
            show_field(field)
            print(f"Победа!!! Выиграл {current_user}")
            break # Теперь этот break точно внутри цикла while

# Проверяем на ничью
        if count == 9:
            show_field(field)
            print("Ничья!")
            break # И этот тоже внутри

# Меняем игрока для следующего хода
        x_turn = not x_turn

# Запуск игры
if __name__ == "__main__":
    start_game()
```
